```python
# Case Study4: E-Commerce Customer Segmentation — Apply K-Means and DBSCAN to an RFM (Recency, Frequency, Monetary) dataset from an online retailer. Reduce dimensions with PCA, visualize segments with t-SNE, and derive actionable marketing personas from each cluster."
```

```python
#sari librabries import kari hai
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.metrics import silhouette_score

import warnings
warnings.filterwarnings('ignore')
```

```python
import kagglehub

# Download latest version data form kaggle
path = kagglehub.dataset_download("datascikhan/e-commerce-customer-segmentation-2026")

print("Path to dataset files:", path)
```

```python
import os

for root, dirs, files in os.walk(path):
    for file in files:
        print(os.path.join(root, file))
```

```python
import pandas as pd
import os

csv_files = []

for root, dirs, files in os.walk(path):
    for file in files:
        if file.endswith('.csv'):
            csv_files.append(os.path.join(root, file))

print(csv_files)
```

```python
df = pd.read_csv(csv_files[0])

print(df.head())
print(df.shape)
print(df.columns.tolist())
print(df.info())
```

```python
df.dropna(subset=['customer_id'], inplace=True)

print(df.head())
print(df.shape)
```

```python
# Define the rfm DataFrame using the dataset's RFM score columns
rfm = df[['recency_score', 'frequency_score', 'monetary_score']]

# Rename columns to standard Recency, Frequency, Monetary names to match downstream cells
rfm = rfm.rename(columns={
    'recency_score': 'Recency',
    'frequency_score': 'Frequency',
    'monetary_score': 'Monetary'
})

rfm_log = np.log1p(rfm)

scaler = StandardScaler()
rfm_scaled = scaler.fit_transform(rfm_log)

print(rfm_scaled[:5])
```

```python
inertia = []

for k in range(2, 11):
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(rfm_scaled)
    inertia.append(km.inertia_)

plt.plot(range(2, 11), inertia, marker='o')
plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.show()
```

```python
kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

rfm['KMeans_Cluster'] = kmeans.fit_predict(rfm_scaled)

print(rfm['KMeans_Cluster'].value_counts())
```

```python
dbscan = DBSCAN(
    eps=0.8,
    min_samples=10
)

rfm['DBSCAN_Cluster'] = dbscan.fit_predict(rfm_scaled)

print(rfm['DBSCAN_Cluster'].value_counts())
```

```python
pca = PCA(n_components=2)

pca_result = pca.fit_transform(rfm_scaled)

plt.figure(figsize=(8, 5))

plt.scatter(
    pca_result[:, 0],
    pca_result[:, 1],
    c=rfm['KMeans_Cluster'],
    cmap='viridis'
)

plt.title('Customer Segmentation using PCA')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.colorbar(label='Cluster')
plt.show()

print("Explained Variance:", pca.explained_variance_ratio_.sum())
```

```python
tsne = TSNE(
    n_components=2,
    random_state=42,
    perplexity=30,
    init='pca',
    learning_rate='auto'
)

tsne_result = tsne.fit_transform(rfm_scaled)

plt.figure(figsize=(8, 5))

plt.scatter(
    tsne_result[:, 0],
    tsne_result[:, 1],
    c=rfm['KMeans_Cluster'],
    cmap='viridis'
)

plt.title('Customer Segmentation using t-SNE')
plt.xlabel('t-SNE 1')
plt.ylabel('t-SNE 2')
plt.colorbar(label='Cluster')
plt.show()
```

```python
kmeans_score = silhouette_score(
    rfm_scaled,
    rfm['KMeans_Cluster']
)

db_labels = rfm['DBSCAN_Cluster']
mask = db_labels != -1

if len(set(db_labels[mask])) > 1:
    dbscan_score = silhouette_score(
        rfm_scaled[mask],
        db_labels[mask]
    )
else:
    dbscan_score = -1

print("K-Means Silhouette Score:", kmeans_score)
print("DBSCAN Silhouette Score:", dbscan_score)
```

```python
personas = rfm.groupby('KMeans_Cluster')[
    ['Recency', 'Frequency', 'Monetary']
].mean()

print(personas)
```
